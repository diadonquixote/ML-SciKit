```python
import pandas as pd
import seaborn as sns

# Step 1: Load the tips dataset
tips = sns.load_dataset("tips")

# Step 2: Extract the target vector (y) and feature matrix (X)
y = tips["tip"]
X = tips.drop(columns=["tip"])

# Step 3: Inspect the shapes
print("Feature matrix (X) shape:", X.shape)
print("Target vector (y) shape:", y.shape)

# Step 4: Preview the features
X.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 5: Convert categorical columns (sex, smoker, day, time) to numeric
X_encoded = pd.get_dummies(X, drop_first=True)

# Step 6: Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# Step 7: Train the model
model = LinearRegression()
model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error

# Step 8: Predict and evaluate
y_pred = model.predict(X_test)
print("R2 Score:", r2_score(y_test, y_pred))
print("RMSE:", root_mean_squared_error(y_test, y_pred))
```
