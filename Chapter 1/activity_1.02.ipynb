```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler

titanic = sns.load_dataset('titanic')

features = ['sex', 'age', 'fare', 'class', 'embark_town', 'alone']
X = titanic[features].copy()
X.head()
```

```python
print(X.isnull().sum())

X['age'] = X['age'].fillna(X['age'].median())
X['embark_town'] = X['embark_town'].fillna(X['embark_town'].mode()[0])

max_fare = X['fare'].mean() + (3 * X['fare'].std())
X['fare'] = X['fare'].clip(upper=max_fare)
```

```python
le = LabelEncoder()
text_cols = ['sex', 'class', 'embark_town', 'alone']

for col in text_cols:
    X[col] = le.fit_transform(X[col].astype(str))

X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Cleaned feature matrix shape:", X_scaled.shape)
```
