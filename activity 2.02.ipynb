```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

data = pd.read_csv("Wholesale customers data.csv")

scaler = StandardScaler()

data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)

data_scaled.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
```

```python
clusters = []
for i in range(1, 11):
    km = KMeans(n_clusters=i, random_state=42)
    km.fit(data_scaled)
    clusters.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(range(1, 11), clusters, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (Average Distance)')
plt.title('Elbow Method For Optimal K')
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(data_scaled)
data_scaled['Cluster'] = kmeans.predict(data_scaled)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.scatter(data_scaled['Grocery'], data_scaled['Milk'], c=data_scaled['Cluster'], cmap='viridis')
ax1.set_xlabel('Grocery')
ax1.set_ylabel('Milk')
ax1.set_title('Clusters: Grocery vs Milk')


ax2.scatter(data_scaled['Frozen'], data_scaled['Detergents_Paper'], c=data_scaled['Cluster'], cmap='viridis')
ax2.set_xlabel('Frozen')
ax2.set_ylabel('Detergents_Paper')
ax2.set_title('Clusters: Frozen vs Detergents_Paper')

plt.tight_layout()
plt.show()
```
